Problem Statement – HR Employee
Performance Dataset
You are provided with a small HR employee dataset containing department,
employee name, salary, and experience.

```python
import pandas as pd

data = {
    "Department": ["HR", "IT", "HR", "IT", "Finance", "Finance"],
    "Employee": ["Asha", "Ravi", "John", "Meera", "Kiran", "Lata"],
    "Salary": [45000, 60000, 52000, 65000, 70000, 68000],
    "Experience": [2, 4, 3, 5, 6, 4]
}

df = pd.DataFrame(data)

print(df)
```

```python
#1. Find the average salary per department.
df.groupby("Department")['Salary'].mean()
```

```python
#2. Count the number of employees in each department.
df.groupby("Department")['Employee'].count()
```

```python
#3. Compute the total experience for each department.
df.groupby("Department")['Experience'].sum()
```

```python
#4. Find the highest salary within each department.
df.groupby("Department")['Salary'].max()
```

```python
#5. Perform multiple aggregations: mean salary and max experience.
df.agg(
    {
        'Salary': 'mean',
        'Experience':'max'
        }
)
```

```python
#6. Sort departments by total salary spent.
df.groupby("Department")['Salary'].sum().sort_values(ascending=False)
```

```python
#7. Obtain min and max salary for every department.
min1=df.groupby("Department")['Salary'].min()
max1=df.groupby("Department")['Salary'].max()
print(min1)
print(max1)
```

```python
#8. Generate descriptive statistics for salary grouped by department.
df.groupby("Department")['Salary'].describe()
```

```python
#9. Display employee count and average experience per department using agg.
df.groupby("Department").agg(
    {
    'Employee': 'count' ,
'Experience': 'mean'

    }


)
```

```python
#Compute salary range (max - min) for each department using a custom function.

df.groupby('Department')['Salary'].agg(
       lambda x: (x.max(),x.min())
)

        
```

```python
# using keyword argument
df.groupby('Department')['Salary'].agg(
    min_salary='min',
    max_salary='max',
    salary_diff=lambda x: x.max() - x.min()
)
```

```python
# use keyword argument
df.groupby('Department')['Salary'].agg(
    
       salary_diff=lambda x: (x.max()-x.min())
)
```

```python
#6. Create a pie chart showing  distribution across departments.
temp1=df['Department'].value_counts()
temp1

plt.pie(temp1.values,labels=temp1.index, autopct='%1.1f%%')
```

```python
#7. Plot a correlation heatmap for salary and experience.
temp3=df[['Salary','Experience']].corr()
plt.figure(figsize=(5,6))
sns.heatmap(temp3, annot=True,linewidth=2 ,cmap='summer')
```

```python
#8. Draw a line plot of employees vs their salaries.
sns.relplot(data=df,x='Employee',y='Salary', kind='line', color='red', marker='*', markersize=11, )
#hue='Department'
```
