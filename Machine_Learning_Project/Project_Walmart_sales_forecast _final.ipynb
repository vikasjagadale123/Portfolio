```python
import pandas as pd
df1=pd.read_csv("train.csv")
df2=pd.read_csv("stores.csv")
df3=pd.read_csv("features.csv")
```

#  Walmart Sales Forecasting using Machine Learning

##  Objective
- Analyze historical sales data
- Build predictive models
- Optimize performance
- Support business decisions

```python
df = df1.merge(df3, on=['Store','Date'], how='left')
df = df.merge(df2, on='Store', how='left')
df.isnull().sum()
```

```python
# Markdown  represent discount / promotion values applied to products.
md_cols = ['MarkDown1','MarkDown2','MarkDown3','MarkDown4','MarkDown5']
df[md_cols] = df[md_cols].fillna(0)
```

- ##  Data Preprocessing
- Convert date
- Feature engineering

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week
df.drop('Date', axis=1, inplace=True)
```

```python
(df['IsHoliday_x']!=df['IsHoliday_y']).sum()
```

```python
df.columns
```

```python
y=df['Weekly_Sales']
x=df.drop('Weekly_Sales',axis=1)
```

```python
# Is IsHoliday used for label encoding?
# NO — it is NOT used for label encoding (in most cases).
#t is already a binary feature (0/1 or True/False).
```

```python
#check whether your dataset is balanced or imbalanced, you first need to understand:
# This is a regression dataset (target = Weekly_Sales)
#So “balanced/imbalanced” like classification does NOT directly apply
```

```python
x=x.drop('IsHoliday_y', axis=1)
```

```python
pd.get_dummies(x,columns=['Type'],dtype=int)
```

```python
x
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
x['Type']=le.fit_transform(x['Type'])
x
```

```python
from sklearn.preprocessing import StandardScaler
st=StandardScaler()
x=st.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
mse= mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

import numpy as np
rmse=round((np.sqrt(mse)),2)
print("mean_squared_error",mse)
print("rmse",rmse, )
print("accuracy is", round(r2*100,2),"%")
```

```python
from sklearn.neighbors import KNeighborsRegressor
model=KNeighborsRegressor()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)

mse= mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

import numpy as np
rmse=round((np.sqrt(mse)),2)
print("mean_squared_error",mse)
print("rmse",rmse, )
print("accuracy is", round(r2*100,2),"%")
```

```python
print(y_test,y_pred)
```

```python

from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor(n_estimators=5)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
mse= mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

import numpy as np
rmse=round((np.sqrt(mse)),2)
print("mean_squared_error",mse)
print("rmse",rmse, )
print("accuracy is", round(r2*100,2),"%")
```

```python

from sklearn.tree import DecisionTreeRegressor
model= DecisionTreeRegressor()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
mse= mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)

import numpy as np
rmse=round((np.sqrt(mse)),2)
print("mean_squared_error",mse)
print("rmse",rmse, )
print("accuracy is", round(r2*100,2),"%")
```
